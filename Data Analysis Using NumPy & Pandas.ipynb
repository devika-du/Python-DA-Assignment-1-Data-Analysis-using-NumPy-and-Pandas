```python
import numpy as np
import pandas as pd
import warnings 
warnings.filterwarnings("ignore")
```

##### Numpy Array Operations

```python
temperatures_w1=np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])
temperatures_w1
```

```python
print(temperatures_w1.shape)
print(temperatures_w1.dtype)
print(temperatures_w1.size)
```

```python
Fahrenheit = (temperatures_w1 * 9/5) + 32
print(Fahrenheit)
```

```python
print(temperatures_w1.max())
print(temperatures_w1.min())
print(temperatures_w1.mean().round(2))
```

##### Array Slicing and Indexing

```python
print(temperatures_w1[0:3])
print(temperatures_w1[-2:])
print(temperatures_w1[2:5])
```

##### Create a 2D Array

```python
temperatures=np.array([[22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],
                      [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]])
temperatures
```

```python
print(temperatures.shape)
print(temperatures.dtype)
print(temperatures.size)
```

```python
print(temperatures[0,-2:]) #week1
print(temperatures[1,-2:]) #week2
```

#### Pandas Series

```python
marks=pd.Series([95, 92, 89, 85, 80 ],index=["Rank1","Rank2","Rank3","Rank4","Rank5"],name="Marks")
marks
```

##### Indexing and Slicing

```python
print(marks["Rank1"])
```

```python
print(marks.loc["Rank1":"Rank3"])
```

```python
print(marks.iloc[2])
```

```python
print([marks>90])
```

##### Manipulating Series

```python
marks["Rank1"]=marks["Rank1"]=100
```

```python
marks=marks.drop(marks.index[-1])
marks
```

```python
CGPA=marks/10
print(CGPA)
```
